```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_absolute_percentage_error
import numpy as np
import warnings

warnings.filterwarnings("ignore")

# Load dataset
df = pd.read_csv("indjob.csv")
df['Date'] = pd.to_datetime(df['Date'], format="%d-%m-%Y")

# Initialize final dataframe for forecasts
final_forecast_df = pd.DataFrame()

# sMAPE function
def smape(y_true, y_pred):
    return 100 * np.mean(2 * np.abs(y_pred - y_true) / (np.abs(y_pred) + np.abs(y_true)))

# MASE function
def mase(training_series, testing_series, prediction_series):
    n = len(training_series)
    d = np.abs(np.diff(training_series)).sum() / (n - 1)
    errors = np.abs(testing_series - prediction_series)
    return errors.mean() / d if d != 0 else np.nan

# Get unique Field-Location combinations
job_field_location = df[['Field', 'Location']].drop_duplicates()

results = {}

for _, row in job_field_location.iterrows():
    field = row['Field']
    location = row['Location']
    print(f"\nProcessing Field: {field} | Location: {location}")

    filtered_df = df[(df['Field'] == field) & (df['Location'] == location)]
    daily_stats = filtered_df.groupby('Date').agg(
        jobs_posted=('Field', 'count'),
        average_salary=('Salary', 'mean')
    ).reset_index().sort_values(by='Date')

    if len(daily_stats) < 10:
        print(f"Skipping {field} | {location}, not enough data.")
        continue

    # Train-test split
    train_size = int(0.7 * len(daily_stats))
    train_jobs = daily_stats.iloc[:train_size]
    test_jobs = daily_stats.iloc[train_size:]

    # ARIMA for jobs
    model_jobs = ARIMA(train_jobs['jobs_posted'], order=(5, 1, 0)).fit()
    pred_jobs = model_jobs.forecast(steps=len(test_jobs))

    # ARIMA for salary
    model_salary = ARIMA(train_jobs['average_salary'], order=(5, 1, 0)).fit()
    pred_salary = model_salary.forecast(steps=len(test_jobs))

    # Metrics
    results[f"{field} | {location}"] = {
        'MAPE_Jobs': mean_absolute_percentage_error(test_jobs['jobs_posted'], pred_jobs),
        'sMAPE_Jobs': smape(test_jobs['jobs_posted'].values, pred_jobs),
        'MASE_Jobs': mase(train_jobs['jobs_posted'].values, test_jobs['jobs_posted'].values, pred_jobs),
        'MAPE_Salary': mean_absolute_percentage_error(test_jobs['average_salary'], pred_salary),
        'sMAPE_Salary': smape(test_jobs['average_salary'].values, pred_salary),
        'MASE_Salary': mase(train_jobs['average_salary'].values, test_jobs['average_salary'].values, pred_salary),
    }

    # Future forecast (30 days ahead)
    full_model_jobs = ARIMA(daily_stats['jobs_posted'], order=(5, 1, 0)).fit()
    full_model_salary = ARIMA(daily_stats['average_salary'], order=(5, 1, 0)).fit()

    future_jobs = full_model_jobs.forecast(steps=30)
    future_salary = full_model_salary.forecast(steps=30)
    future_dates = pd.date_range(daily_stats['Date'].max() + pd.Timedelta(days=1), periods=30)

    forecast_df = pd.DataFrame({
        'Date': future_dates,
        'Field': field,
        'Location': location,
        'Jobs_Predicted': future_jobs,
        'Average_Salary_Predicted': future_salary
    })

    final_forecast_df = pd.concat([final_forecast_df, forecast_df], ignore_index=True)

    # Plot for jobs posted
    plt.figure(figsize=(10, 4))
    plt.plot(train_jobs['Date'], train_jobs['jobs_posted'], label="Train Jobs", color='blue')
    plt.plot(test_jobs['Date'], test_jobs['jobs_posted'], label="Actual Jobs", color='green')
    plt.plot(test_jobs['Date'], pred_jobs, label="Forecasted Jobs", color='red', linestyle='--')
    plt.title(f"{field} | {location} - Jobs Forecast")
    plt.xlabel("Date")
    plt.ylabel("Jobs Posted")
    plt.legend()
    plt.axvline(test_jobs['Date'].iloc[0], color='black', linestyle='dashed')
    plt.tight_layout()
    plt.show()

    # Plot for average salary
    plt.figure(figsize=(10, 4))
    plt.plot(train_jobs['Date'], train_jobs['average_salary'], label="Train Salary", color='purple')
    plt.plot(test_jobs['Date'], test_jobs['average_salary'], label="Actual Salary", color='orange')
    plt.plot(test_jobs['Date'], pred_salary, label="Forecasted Salary", color='brown', linestyle='--')
    plt.title(f"{field} | {location} - Salary Forecast")
    plt.xlabel("Date")
    plt.ylabel("Average Salary")
    plt.legend()
    plt.axvline(test_jobs['Date'].iloc[0], color='black', linestyle='dashed')
    plt.tight_layout()
    plt.show()

# Save the 30-day forecasts to CSV
final_forecast_df.to_csv("30_day_job_salary_forecast.csv", index=False)

# Print summary metrics
print("\n--- Forecast Evaluation Metrics ---")
all_mape_jobs = []
all_smape_jobs = []
all_mase_jobs = []
all_mape_salary = []
all_smape_salary = []
all_mase_salary = []

for key, val in results.items():
    print(f"\n{key}:")
    print(f"  Jobs   → MAPE: {val['MAPE_Jobs']:.2%}, sMAPE: {val['sMAPE_Jobs']:.2f}%, MASE: {val['MASE_Jobs']:.3f}")
    print(f"  Salary → MAPE: {val['MAPE_Salary']:.2%}, sMAPE: {val['sMAPE_Salary']:.2f}%, MASE: {val['MASE_Salary']:.3f}")
    all_mape_jobs.append(val['MAPE_Jobs'])
    all_smape_jobs.append(val['sMAPE_Jobs'])
    all_mase_jobs.append(val['MASE_Jobs'])
    all_mape_salary.append(val['MAPE_Salary'])
    all_smape_salary.append(val['sMAPE_Salary'])
    all_mase_salary.append(val['MASE_Salary'])

# Averages
print("\n--- Average Forecast Errors Across All Fields & Locations ---")
print(f"Jobs   → Avg MAPE: {np.mean(all_mape_jobs):.2%}, Avg sMAPE: {np.mean(all_smape_jobs):.2f}%, Avg MASE: {np.mean(all_mase_jobs):.3f}")
print(f"Salary → Avg MAPE: {np.mean(all_mape_salary):.2%}, Avg sMAPE: {np.mean(all_smape_salary):.2f}%, Avg MASE: {np.mean(all_mase_salary):.3f}")
```
