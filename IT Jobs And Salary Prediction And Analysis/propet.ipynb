```python
import pandas as pd
import matplotlib.pyplot as plt
from prophet import Prophet
from sklearn.metrics import mean_absolute_percentage_error
import numpy as np
import warnings

warnings.filterwarnings("ignore")

# Load dataset
df = pd.read_csv("indjob.csv")
df['Date'] = pd.to_datetime(df['Date'], format="%d-%m-%Y")

final_forecast_df = pd.DataFrame()

# sMAPE
def smape(y_true, y_pred):
    return 100 * np.mean(2 * np.abs(y_pred - y_true) / (np.abs(y_pred) + np.abs(y_true)))

# MASE
def mase(training_series, testing_series, prediction_series):
    n = len(training_series)
    d = np.abs(np.diff(training_series)).sum() / (n - 1)
    errors = np.abs(testing_series - prediction_series)
    return errors.mean() / d if d != 0 else np.nan

job_field_location = df[['Field', 'Location']].drop_duplicates()
results = {}

for _, row in job_field_location.iterrows():
    field, location = row['Field'], row['Location']
    print(f"\nProcessing Field: {field} | Location: {location}")

    filtered_df = df[(df['Field'] == field) & (df['Location'] == location)]
    daily_stats = filtered_df.groupby('Date').agg(
        jobs_posted=('Field', 'count'),
        average_salary=('Salary', 'mean')
    ).reset_index().sort_values(by='Date')

    if len(daily_stats) < 10:
        print(f"Skipping {field} | {location}, not enough data.")
        continue

    train_size = int(0.7 * len(daily_stats))
    train_df, test_df = daily_stats[:train_size], daily_stats[train_size:]

    # Prophet for jobs
    jobs_train = train_df[['Date', 'jobs_posted']].rename(columns={'Date': 'ds', 'jobs_posted': 'y'})
    model_jobs = Prophet()
    model_jobs.fit(jobs_train)
    jobs_future = test_df[['Date']].rename(columns={'Date': 'ds'})
    pred_jobs = model_jobs.predict(jobs_future)['yhat'].values

    # Prophet for salary
    salary_train = train_df[['Date', 'average_salary']].rename(columns={'Date': 'ds', 'average_salary': 'y'})
    model_salary = Prophet()
    model_salary.fit(salary_train)
    salary_future = test_df[['Date']].rename(columns={'Date': 'ds'})
    pred_salary = model_salary.predict(salary_future)['yhat'].values

    results[f"{field} | {location}"] = {
        'MAPE_Jobs': mean_absolute_percentage_error(test_df['jobs_posted'], pred_jobs),
        'sMAPE_Jobs': smape(test_df['jobs_posted'].values, pred_jobs),
        'MASE_Jobs': mase(train_df['jobs_posted'].values, test_df['jobs_posted'].values, pred_jobs),
        'MAPE_Salary': mean_absolute_percentage_error(test_df['average_salary'], pred_salary),
        'sMAPE_Salary': smape(test_df['average_salary'].values, pred_salary),
        'MASE_Salary': mase(train_df['average_salary'].values, test_df['average_salary'].values, pred_salary),
    }

    # Forecast next 30 days
    future_dates = pd.date_range(daily_stats['Date'].max() + pd.Timedelta(days=1), periods=30)

    jobs_future_df = pd.DataFrame({'ds': future_dates})
    jobs_forecast = model_jobs.predict(jobs_future_df)['yhat']

    salary_future_df = pd.DataFrame({'ds': future_dates})
    salary_forecast = model_salary.predict(salary_future_df)['yhat']

    forecast_df = pd.DataFrame({
        'Date': future_dates,
        'Field': field,
        'Location': location,
        'Jobs_Predicted': jobs_forecast,
        'Average_Salary_Predicted': salary_forecast
    })

    final_forecast_df = pd.concat([final_forecast_df, forecast_df], ignore_index=True)

    # Plot jobs
    plt.figure(figsize=(12, 5))
    plt.plot(train_df['Date'], train_df['jobs_posted'], label="Train Jobs", color='blue')
    plt.plot(test_df['Date'], test_df['jobs_posted'], label="Actual Jobs", color='green')
    plt.plot(test_df['Date'], pred_jobs, label="Forecasted Jobs", color='red', linestyle='--')
    plt.title(f"{field} | {location} - Jobs Forecast")
    plt.xlabel("Date")
    plt.ylabel("Jobs Posted")
    plt.legend()
    plt.tight_layout()
    plt.show()

    # Plot salary
    plt.figure(figsize=(12, 5))
    plt.plot(train_df['Date'], train_df['average_salary'], label="Train Salary", color='purple')
    plt.plot(test_df['Date'], test_df['average_salary'], label="Actual Salary", color='orange')
    plt.plot(test_df['Date'], pred_salary, label="Forecasted Salary", color='brown', linestyle='--')
    plt.title(f"{field} | {location} - Salary Forecast")
    plt.xlabel("Date")
    plt.ylabel("Average Salary")
    plt.legend()
    plt.tight_layout()
    plt.show()

# Save 30-day forecast
final_forecast_df.to_csv("30_day_job_salary_forecast_prophet.csv", index=False)

# Summary metrics
print("\n--- Forecast Evaluation Metrics ---")
all_mape_jobs, all_smape_jobs, all_mase_jobs = [], [], []
all_mape_salary, all_smape_salary, all_mase_salary = [], [], []

for key, val in results.items():
    print(f"\n{key}:")
    print(f"  Jobs   → MAPE: {val['MAPE_Jobs']:.2%}, sMAPE: {val['sMAPE_Jobs']:.2f}%, MASE: {val['MASE_Jobs']:.3f}")
    print(f"  Salary → MAPE: {val['MAPE_Salary']:.2%}, sMAPE: {val['sMAPE_Salary']:.2f}%, MASE: {val['MASE_Salary']:.3f}")
    all_mape_jobs.append(val['MAPE_Jobs'])
    all_smape_jobs.append(val['sMAPE_Jobs'])
    all_mase_jobs.append(val['MASE_Jobs'])
    all_mape_salary.append(val['MAPE_Salary'])
    all_smape_salary.append(val['sMAPE_Salary'])
    all_mase_salary.append(val['MASE_Salary'])

# Print average errors
print("\n--- Average Forecast Errors ---")
print(f"Jobs   → Avg MAPE: {np.mean(all_mape_jobs):.2%}, Avg sMAPE: {np.mean(all_smape_jobs):.2f}%, Avg MASE: {np.mean(all_mase_jobs):.3f}")
print(f"Salary → Avg MAPE: {np.mean(all_mape_salary):.2%}, Avg sMAPE: {np.mean(all_smape_salary):.2f}%, Avg MASE: {np.mean(all_mase_salary):.3f}")
```
